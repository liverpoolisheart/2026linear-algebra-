```python
#include <stdio.h>
#include <stdlib.h>


int n;
int** arr;
int cnt[3];

int idx2n(int idx){
    return idx + 1; //idx= -1이면 cnt[0] 이렇게 되게 할려고.
}

int checkSameCells(int xs,int xe,int ys,int ye){
    int i,j,idx;
    bool flag = true; //모든 원소가 같은지 나타내는 변수.flag =
    int mycnt=0; // 검사한 원소의 개수를 세는 변수

    idx = arr[xs][ys];
    for ( i = xs;i <=xe;i++){
        for (j=ys;j<=ye;j++){
            if (idx != arr[i][j]){
                flag = false;
                break;
            }
            mycnt++;
        }
    }
    return flag;
}


void dnc(int xs,int xe,int ys,int ye) //재귀분할을 이용한 정복.
{
    int i,j,idx;

    idx=arr[xs][ys];
    // 더이상 나눌 수 없는 가장 작은 영역에 도달햇을 때 재귀를 끝내는 조건.
    if (xs==xe&&ys==ye){ //시작행과 끝행이 같고, 시작열과 끝 열이 같다면 = 현재 원소가 하나인지 확인
        cnt[idx2n(idx)]++; // ex) cnt[1]++ 이면 cnt[1]의 값을 1증가/ ++은 변수 뿐만 아니라 배열도 된다.
        return;
    }

    bool flag = checkSameCells(xs,xe,ys,ye);
    if (flag){
        cnt[idx2n(idx)]++; //첫바퀴 때는 idx= 전체 영역의 첫번째 원소
        return; //현재 dnc 종료.
    }

    //분할 정복의 핵심(영역을 9개로 나누기 위한 좌표계산)
    int xm1,xm2,ym1,ym2;
    // 2를 곱하는 이유는 계산 맞출려고
    xm1=(2*xs+xe)/3; //인덱스를 통해 행을 3등분 내기: 1/3 지점
    xm2=(xs+2*xe)/3; // 2/3 지점
    ym1=(2*ys+ye)/3; // y축의 1/3지점
    ym2=(ys+2*ye)/3; // y축의 2/3지점

    // 9개의 dnc()호출
    dnc(xs,xm1,ys,ym1);
    dnc(xm1+1,xm2,ys,ym1);
    dnc(xm2+1,xe,ys,ym1);
    dnc(xs,xm1,ym1+1,ym2);
    dnc(xm1+1,xm2,ym1+1,ym2);
    dnc(xm2+1,xe,ym1+1,ym2);
    dnc(xs,xm1,ym2+1,ye);
    dnc(xm1+1,xm2,ym2+1,ye);
    dnc(xm2+1,xe,ym2+1,ye);
}

int main(){
    int i,j,val;

    for (i = 0; i<3;i++) cnt[i]=0;

    scanf("%d",&n);
    arr = (int**)malloc(sizeof(int*)*n);
    for ( i = 0 ; i < n ; i++){
        arr[i]=(int*)malloc(sizeof(int)*n);
        for (j=0;j<n;j++){ // N*N 배열 노가다로 채워넣기
            scanf("%d",&val);
            arr[i][j]=val;
        }
    }

    dnc(0,n-1,0,n-1);

    for (i=0 ; i < 3; i++) printf("%d\n",cnt[i]);
    for ( i = 0; i < n ;i++) free(arr[i]); //free안에는 할당받은 메모리 주소를 넣어주면 됨.
    free(arr); //그다음 arr자체의 메모리를 해제

    return 0;
}
```
